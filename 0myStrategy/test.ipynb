```python
import requests
import pandas as pd
from collections import defaultdict
```

```python
# دریافت تاریخچه قیمت
PRICE_URL = "https://cdn.tsetmc.com/api/ClosingPrice/GetClosingPriceDailyList/7535706950108842/0"
HEADERS  = {'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64; rv:157.0) Gecko/20100101 Firefox/15'}

response = requests.get(PRICE_URL, headers= HEADERS, timeout= 3)
response.text
raw_data = response.json().get("closingPriceDaily")
df = pd.json_normalize(raw_data)
df['dEven'] = df['dEven'].astype(str)
df['date'] = pd.to_datetime(df['dEven'], format='%Y%m%d')
df = df.sort_values('date').reset_index(drop=True)
df
```

```python
# ریز معاملات
PRICE_URL = "https://cdn.tsetmc.com/api/Trade/GetTrade/53874633519758686"
HEADERS  = {'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36'}

try:
    response = requests.get(PRICE_URL, headers= HEADERS, timeout= 10)
    if response.status_code == 200:
        raw_data = response.json().get("trade")
        df = pd.json_normalize(raw_data)
        df['time'] = (df['hEven'].astype(int).astype(str).str.zfill(6).pipe(pd.to_datetime, format='%H%M%S').dt.time)
        df = df.sort_values('time').reset_index(drop=True)
        df.to_excel('718.xlsx')

        price_dict = defaultdict(list)
        for h, p in zip(df['hEven'], df['pTran']):
            price_dict[h].append(p)

        price_dict = dict(price_dict)
        print(price_dict)
    else:
        print('No Data')
except Exception as e:
    print(f"error response: {e}")
```
